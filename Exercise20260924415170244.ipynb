````python
# 習題：數學物件、Python 型別與 AI 中的數值表示

在數學中，我們經常接觸下列概念：

- 自然數（Natural Numbers）
- 整數（Integers）
- 有理數（Rational Numbers）
- 實數（Real Numbers）
- 複數（Complex Numbers）
- 向量（Vectors）
- 矩陣（Matrices）
- 集合（Sets）
- 命題的真假（Truth Values）

本習題將探討：

```text
數學物件
    ↓
Python 型別
    ↓
CPython 實作
    ↓
AI 中的數值表示
```

---

## 第一部分：Python 如何表示數學物件？

請查閱 Python 文件或自行測試，完成下表。

| 數學概念 | Python 型別 | 範例 |
|----------|------------|------|
| 自然數 | int | 5 |
| 整數 | int | -10 |
| 有理數 | fractions.Fraction | Fraction(3, 4) |
| 實數 | float | 3.14159 |
| 複數 | complex | 2 + 3j |
| 向量 | list、tuple或numpy.ndarray | [1, 2, 3] |
| 矩陣 | 巢狀list或numpy.ndarray | [[1, 2], [3,4]] |
| 集合 | set | {1, 2, 3} |
| 命題真假 | bool | True、False |

並回答下列問題：

1. 哪些是 Python 的內建型別（Built-in Types）？int、float、complex、bool、list、tuple、set
2. 哪些需要額外模組才能較完整地表示？有理數、向量與矩陣
3. Python 的型別分類與數學中的分類是否完全一致？不完全一致

---

## 第二部分：數學概念與 Python 型別

數學上有下列包含關係：

$$
\mathbb N
\subseteq
\mathbb Z
\subseteq
\mathbb Q
\subseteq
\mathbb R
\subseteq
\mathbb C
$$

請回答：

1. Python 是否為上述每一個數系都提供獨立型別？否
2. 為什麼自然數與整數都使用 `int` 表示？因為自然數是整數的子集合
3. Python 為什麼提供 `complex` 型別？因為許多科學與工程問題需要直接使用複數
4. Python 為什麼沒有內建的有理數型別？使用需求較少、計算成本較高、需要時仍可使用標準函式庫

5. 下列三個物件的型別有何不同？數值相同，但型別不同

```python
1
1.0
1+0j
```

---

## 第三部分：CPython 如何實作這些型別？

除了知道 Python 使用哪些型別表示數學物件之外，

我們也希望了解：

> Python 物件在 CPython 中是如何實作的？

請查閱資料，完成下表。

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|
| 自然數 | `int` | PyLongObject | 任意經度整數(Unlimited Precision) |
| 整數 | `int` | PyLongObject | 不受32-bit或64-bit限制，可自動擴展位數 |
| 實數 | `float` | PyFloatObject | 採用lEEE 754雙精度浮點數(double) |
| 複數 | `complex` | PyComplexObject | 由兩個double組成(實部與虛部) |
| 集合 | `set` | PySetObject | 以雜湊表(Hash Table)實作，元素不可重複 |
| 命題真假 | `bool` | PyBoolObject | bool是int的子型別，只有True與False兩個實例 |
| 串列 | `list` | PyListObject | 動態陣列(Dynamic Array)，支援快速索引 |

---

## 第四部分：深入探討各種型別

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？
2. 對應的 CPython 物件結構名稱為何？
3. 它如何儲存資料？
4. 是否有固定大小限制？
5. 若資料規模持續增加，如何擴充儲存空間？

| 型別 | 內健型別 | CPython結構名稱 |　如何儲存資料　｜　固定大小限制　｜　如何擴充空間　｜
｜ｉｎｔ｜PyLongObject｜拆成多個區塊(digits)儲存整數｜否｜配置更多digits陣列空間｜
｜ｆｌｏａｔ｜是｜ PyFloatObject｜一個Ｃ　ｄｏｕｂｌｅ|是|無法擴充，固定格式|
|complex|是|PyComplexObject|兩個 C double（實部、虛部）|是|無法擴充，固定格式|
|bool|是|PyBoolObject|本質上繼承 int，只有 True/False|是|不需要擴充|
|set|是|PySetObject|Hash Table（雜湊表）|否|負載過高時重新配置更大的表格|
|list|是|PyListObject|Dynamic Array（動態陣列）|否|預留額外容量，不足時重新配置更大的陣列|

---

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為Python的int不是固定的整數型別
2. Python 的 `int` 是否具有固定的位元數限制？否
3. 什麼是 Arbitrary Precision（任意精度）？整數的位數不是事先固定的，而是依據資料大小動態成長。
4. CPython 使用什麼物件來實作 `int`？PyLongObject
5. Python 整數大小真正的限制是什麼？電腦的記憶體、作業系統限制與可接受的運算時間

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？電腦內部使用二進位儲存浮點數
2. 浮點數是否能精確表示所有小數？否
3. 什麼是浮點數的精度（precision）？能夠保留多少有效數字
4. 什麼是浮點數的範圍（range）？能表示的最大與最小數值
5. 精度與範圍有何不同？精度:數字有多準，範圍:數值有多大或多小

---

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？FP32、FP16、BF16、FP64、INT8、FP8
2. 什麼是 FP64？64-bit位元數，用於科學計算、高精度數值分析
3. 什麼是 FP32？32-bit，傳統深度學習訓練
4. 什麼是 FP16？16-bit，混合精度訓練、推論
5. 什麼是 BF16？現代大型AI模型訓練主流
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？權重本質上是實數、梯度下降需要小數、Python的int記憶體成本高、AI硬體專門為浮點數設計

---

## 第七部分：比較不同數值表示方式

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int | 不固定 | 否 | 整數運算、計數、索引值 |
| FP64 | 64-bit | 有 | 科學計算、高精度數值分析 |
| FP32 | 32-bit | 有 | 傳統深度學習訓練標準格式 |
| FP16 | 16-bit | 有 | 混合精度訓練、推論加速 |
| BF16 | 16-bit | 有 | 大型AI模型訓練主流 |

並回答：

1. 哪種表示方式最精確？FP64
2. 哪種表示方式最節省記憶體？FP16、BF16
3. 哪種表示方式最常見於 AI 訓練？BF16
4. 哪種表示方式最常見於 AI 推論（Inference）？FP16

---

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？Python 的 int不會被限制在固定的位元數內，而是可以根據數值大小動態增加儲存空間
2. 它是否代表可以儲存無限大的整數？Python可以儲存任意大的有限整數，但不能儲存真正無限大的整數
3. CPython 如何實現這項特性？使用 PyLongObject

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？Python 的 int 底層由 PyLongObject 實作，採用可變長度儲存方式，而不是固定的 32 位元或 64 位元整數
2. 為什麼 `float` 卻存在範圍限制？Python 的 float 通常使用 C 語言的 double，也就是 IEEE 754 的 FP64（雙精度浮點數）。
3. 為什麼 `float` 可能產生誤差？float 用固定 64 位元換取高速運算與廣大範圍，因此會有範圍限制與精度誤差

---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。Python:正確性優先、AI模型常用的FP16、BF16:效率優先
2. 各自的優點與缺點。
Python:完全精確、不會發生一般整數溢位、適合數學與密碼學、記憶體需求高、運算速度較慢
FP16/BF16:記憶體占用小、GPU運算快、可訓練更大的模型、有誤差、精度有限
3. 為什麼大型語言模型通常選擇較低精度的表示方式？模型太大
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？記憶體暴增、訓練速度下降、硬體成本上升、精度提升有限

---

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？因為AI的核心需求不是完全精確，而是在可接受的誤差內使用最少記憶體獲得最高運算速度充分獲得GPU

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。
````
